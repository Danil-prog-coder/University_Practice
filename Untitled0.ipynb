```python
import numpy as np
import operator
import pandas as pd # Для работы с данными
import matplotlib.pyplot as plt  # Библиотека для визуализации результатов
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
```

```python
data = pd.read_csv('data_moscow (1).csv', delimiter=',')
# columns = ['Renovation', 'Region', 'Metro station', 'Apartment type']

# for col in columns:
#     if col in data.columns:
#         print(f"\n=== {col} ===")
#         print(data[col].dropna().unique())
#     else:
#         print(f"\n❌ Колонка '{col}' не найдена в файле")

# =========================
# 1. Region (по качеству)
# =========================

region_map = {
    'Moscow': 1,
    'Moscow region': 2
}
data['Region'] = data['Region'].map(region_map)

# =========================
# 2. RENOVATION (по качеству)
# =========================

renovation_map = {
    'Without renovation': 0,
    'Cosmetic': 1,
    'European-style renovation': 2,
    'Designer': 3
}
data['Renovation'] = data['Renovation'].map(renovation_map)

# =========================
# 3. Apartment type
# =========================

region_map = {
    'New building': 1,
    'Secondary': 2
}
data['Apartment type'] = data['Apartment type'].map(region_map)

# =========================
# 4. Metro station
# =========================

central_metro = {
    'Китай-город', 'Арбатская', 'Тверская', 'Охотный ряд',
    'Пушкинская', 'Театральная', 'Лубянка', 'Кузнецкий мост',
    'Кропоткинская', 'Боровицкая', 'Чистые пруды',
    'Третьяковская', 'Новокузнецкая', 'Парк культуры',
    'Смоленская', 'Киевская', 'Краснопресненская',
    'Павелецкая', 'Комсомольская', 'Белорусская',
    'Маяковская', 'Новослободская'
}

mid_metro = {
    'ВДНХ', 'Алексеевская', 'Динамо', 'Аэропорт',
    'Сокол', 'Беговая', 'ВДНХ', 'Фрунзенская',
    'Спортивная', 'Улица 1905 года', 'Белорусская',
    'Курская', 'Таганская', 'Серпуховская', 'Бауманская'
}

def metro_score(x):
    if pd.isna(x):
        return 3  # неизвестное — среднее значение

    x = x.strip()

    if x in central_metro:
        return 1
    elif x in mid_metro:
        return 2
    else:
        return 3  # окраина / Подмосковье / МЦД / новые станции


if 'Metro station' in data.columns:
    data['Metro station'] = data['Metro station'].apply(metro_score)

train, test = train_test_split(data, test_size=0.2, random_state=42)
data
```

```python
list_columns_train = list(train.columns)
list_columns_train.remove("Price")
X_train = train.loc[:, list_columns_train]
Y_train = train.loc[:, ["Price"]]
X_train.head()
```

```python
list_columns_test = list(test.columns)
list_columns_test.remove("Price")
X_test = test.loc[:, list_columns_test]
Y_test = test.loc[:, ["Price"]]
X_test.head()
```

```python

model = LinearRegression()
model.fit(X_train, Y_train) # метод фит - метод обучения модели
y_predict = model.predict(X_test) # получаем предсказанное значение
y_predict[:10]
```

```python
plt.figure(figsize=(12,9))
plt.plot(X_train, Y_train, 'o')
plt.plot(X_train, y_predict, 'k-')
plt.show()
```
